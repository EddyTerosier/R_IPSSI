1. Créer un vecteur "v" qui comporte les valeurs suivante : -1, 2, 4, 2, 0 et un vecteur "w" qui comporte : 0, 3, -1, 0, 2 ? Calculer la somme, la différence, le produit et le rapport de v et w  ? Calculer v + 3, la somme des éléments de v et leur produit ?

```r
v <- c(-1, 2, 4, 2, 0)
w <- c(0, 3, -1, 0, 2)

v = -1   2   4   2   0
w =  0   3  -1   0   2
```

```r
#Somme
v + w
# [1] -1  5  3  2  2

#différence
v - w
# [1] -1 -1  5  2 -2

#Produit
v * w
# [1]  0  6 -4  0  0

#V/W
v / w
# [1] -Inf 0.6666667 -4. Inf 0.0

#V+3
v + 3
# [1] 2 5 7 5 3

#somme des éléments
sum(v)
# [1] 7

#Produitt des éléments
prod(v)
# [1] 0
```
